#### 10.1 Displaying the names of the unique launch sites  in the space mission


```python
%%sql
select distinct(Launch_site) as LaunchSite from Spacextable order by LaunchSite;
```

####  10.2 Displaying 5 records where launch sites begin with the string 'CCA' 


```python
%%sql 
select * from Spacextable where Launch_site like "CCA%" Limit 5;
```


#### 10.3 Displaying the total payload mass carried by boosters launched by NASA (CRS)


```python
%%sql 
select SUM(PAYLOAD_MASS__KG_) as Total_Payload_Mass from Spacextable where Customer like "%CRS%"
```

#### 10.4 Displaying average payload mass carried by booster version F9 v1.1


```python
%%sql 
select Round(AVG(PAYLOAD_MASS__KG_), 3) as Average_Payload_Mass from Spacextable where Booster_Version like "F9 v1.1%"
```


#### 10.5 Listing the date when the first succesful landing outcome in ground pad was acheived.



```python
%%sql
select distinct(Landing_Outcome) from Spacextable;
```

```python
%%sql 
Select Min(Date) as Min_Date, Landing_Outcome from Spacextable where Landing_Outcome like "Success (ground pad)";
```


#### 10. 6 Listing the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
%%sql
select Booster_Version, Payload, PAYLOAD_MASS__KG_, Landing_Outcome 
from Spacextable 
where Landing_Outcome like "Success (drone ship)" AND
PAYLOAD_MASS__KG_ between 4000 and 6000;
```


#### 10.7 Listing the total number of successful and failure mission outcomes


```python
%%sql 
select Mission_OUtcome, Count(Mission_Outcome) as Total_Number from Spacextable 
Group By Mission_Outcome;
```


#### 10.8 Listing the   names of the booster_versions which have carried the maximum payload mass.


```python
%%sql
select Booster_Version, PAYLOAD_MASS__KG_ from Spacextable
where PAYLOAD_MASS__KG_ = (Select Max(PAYLOAD_MASS__KG_) from Spacextable);
```


#### 10.9 Listing the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.



Used  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.

```python
%%sql 
Select substr(Date, 6, 2) as Month, Date, Booster_Version, Launch_Site, Landing_Outcome
from Spacextable
where Landing_Outcome like "Failure (drone ship)" AND
substr(Date, 0, 5) = "2015";
```

#### 10.10 Ranking the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20 in descending order.


```python
%%sql 
select Landing_Outcome, count(Landing_Outcome) as Count_outcome, Date
from spacextable
where Date between "2010-06-04" and "2017-03-20"
Group By Landing_Outcome
order by Count_outcome DESC;
```
